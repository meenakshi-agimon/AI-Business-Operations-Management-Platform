There is a strong positive relationship between workload percentage and the number of active tasks. This means that employees with more active tasks generally tend to have higher workloads.

- Estimated Hours and Hours Logged — Strong Positive Correlation (0.69)

Allocation score has a strong positive relationship with skill match. Records with a matching employee skill tend to have higher allocation scores.

There is a moderate positive relationship between hours logged and task progress. Tasks with more hours logged generally tend to show greater progress.

- Experience and Performance — Moderate Positive Correlation (0.49)
Employee experience has a moderate positive relationship with performance score. More experienced employees generally tend to have somewhat higher performance scores.

- Allocation Score and Recommendation — Moderate Positive Correlation (0.37)
Allocation score has a positive relationship with employee recommendation. Higher allocation scores are generally associated with a greater likelihood of an employee being recommended.

Skill match has a positive relationship with employee recommendation. Employees whose skills match the task requirements are somewhat more likely to be recommended.

## PART 7 — Outlier Analysis
### 7.1 Select the important numerical variables

```python
outlier_features = [
    'experience_years',
    'workload_percentage',
    'active_tasks',
    'performance_score',
    'estimated_hours',
    'hours_logged',
    'progress_percentage',
    'allocation_score'
]
```

```python
outlier_summary = []

for feature in outlier_features:
    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[feature] < lower_bound) |
        (df[feature] > upper_bound)
    ]
    
    outlier_summary.append({
        'Feature': feature,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers),
        'Outlier Percentage': (len(outliers) / len(df)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
outlier_summary.sort_values(
    by='Outlier Count',
    ascending=False
)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=outlier_summary,
    x='Feature',
    y='Outlier Count'
)

plt.title('Number of Potential Outliers by Feature')
plt.xlabel('Feature')
plt.ylabel('Number of Potential Outliers')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## PART 8 — BUSINESS INSIGHTS

1. Workload Management

The analysis shows a strong relationship between workload percentage and active tasks (0.74). Employees with more active tasks generally have higher workloads. Therefore, the system should consider an employee's current workload and number of active tasks before assigning additional work.

2. Skill-Based Employee Allocation

Only around 20% of records have a skill match, while the majority do not. This indicates that finding an employee with the required skills can be a challenge. The allocation system should therefore give importance to matching employee skills with task requirements.

3. Skill Match and Allocation Score

There is a strong positive relationship (0.70) between skill match and allocation score. This indicates that employees whose skills match the task requirements tend to receive higher allocation scores. Therefore, skill compatibility is an important factor when determining employee suitability.

4. Allocation Score and Recommendation

Allocation score has a positive relationship (0.37) with employee recommendation. Employees with higher allocation scores tend to be recommended more often. This suggests that the allocation score can be useful for ranking suitable employees for a task.

5. Workload and Recommendation

Workload has a weak negative relationship (-0.13) with employee recommendation. This suggests that employees with higher workloads are slightly less likely to be recommended for additional tasks. Therefore, the system should avoid assigning new tasks to employees who are already heavily loaded whenever suitable alternatives are available.

6. Employee Availability

Most employees are marked as Available, while a significant portion are Busy and a smaller number are On Leave. This shows that availability is an important practical factor in task allocation. An employee may have the required skills but still not be suitable if they are unavailable or already busy.

7. Employee Experience and Performance

Experience has a moderate positive relationship (0.49) with performance score. More experienced employees generally tend to have somewhat higher performance scores. Experience can therefore be considered as one supporting factor when evaluating employee suitability, rather than relying on experience alone.

8. Task Priority and Status

Most tasks are Medium or High priority, while Critical tasks are less common. Most tasks are currently In Progress or Completed, with fewer tasks being Not Started or Blocked. This suggests that task allocation should consider both priority and current task status so that important tasks receive appropriate resources.

9. Task Hours and Workload Planning

Estimated hours and hours logged have a strong positive relationship (0.69). Tasks requiring more estimated hours generally tend to have more hours logged. This indicates that estimated working hours can provide useful information for workload planning and resource allocation.

10. Task Progress

The average task progress is approximately 79%, and many tasks have progress between 80–100%. This indicates that most tasks in the dataset are relatively well progressed. However, low-progress tasks should be investigated, particularly when they are associated with Blocked or Not Started statuses.

11. Outliers Require Investigation

The outlier analysis identified unusual observations, particularly in hours logged, experience, progress, estimated hours, and allocation score. These should not automatically be removed because some extreme values may represent genuine business situations. For example, high logged hours may indicate a complex task.

The EDA shows that effective employee-task allocation requires consideration of multiple factors rather than relying on a single variable. Workload and active tasks are strongly related, indicating that current workload is important when assigning new tasks. Skill matching has a strong relationship with allocation score, while allocation score and skill matching are positively associated with employee recommendation. Employee availability, experience, performance, task priority, task status, and estimated working hours also provide useful information for making allocation decisions. The large proportion of records without a direct skill match highlights the need for better skill-based employee-task matching. Overall, the findings support the development of a data-driven employee allocation system that balances skill suitability, workload, availability, performance, experience, and task requirements to recommend appropriate employees for tasks.
