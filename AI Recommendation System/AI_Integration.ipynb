```python
def recommend_employee(row):

    emp = row.copy()

    emp = emp.drop(
        labels=["employee_id", "recommended_employee"]
    )

    emp_df = pd.DataFrame([emp])

    prediction = model.predict(emp_df)[0]

    probability = model.predict_proba(emp_df)[0][1]

    return prediction, probability
```

```python
for i in range(5):

    employee = df.iloc[i]

    prediction, probability = recommend_employee(employee)

    print(
        employee["employee_id"],
        prediction,
        probability
    )
```

```python
# Top recommended employees
```

```python
results = []

for i in range(len(df)):

    prediction, probability = recommend_employee(df.iloc[i])

    results.append([
        df.iloc[i]["employee_id"],
        prediction,
        probability
    ])

result_df = pd.DataFrame(
    results,
    columns=[
        "employee_id",
        "recommended",
        "probability"
    ]
)
```

```python
top_employees = result_df.sort_values(
    by="probability",
    ascending=False
)

top_employees.head(10)
```

```python
def get_best_employees(data, top_n=5):

    temp = []

    for i in range(len(data)):

        prediction, probability = recommend_employee(data.iloc[i])

        if prediction == 1:

            temp.append([
                data.iloc[i]["employee_id"],
                probability
            ])

    result = pd.DataFrame(
        temp,
        columns=["employee_id", "probability"]
    )

    return result.sort_values(
        by="probability",
        ascending=False
    ).head(top_n)
```
